```python
#---------------------------------------------
#Creating a new column based on Month &Year
#---------------------------------------------

df['Date'] = pd.to_datetime(
    df['Year'].astype(str) + '-' +
    df['Month'].astype(str) + '-01')

df[['Year', 'Month', 'Date']].head()
print(df[['Year','Month','Date']].head(5))   # first 5 rows
print(df[['Year','Month','Date']].tail(5))   # last 5 rows
```

```python
#Creating Derived Columns
#---------------------------------------------
#Waste Processing Efficiency
#---------------------------------------------

df['Processing_Efficiency'] = (df['Waste_Processing_Capacity'] /df['Daily_Waste_Generation_Tons']) * 100
print("Processing_Efficiency:",df['Processing_Efficiency'])
```

```python
#Shows how much of the generated waste is actually processed. Helps measure system efficiency.
#City officials can identify if processing plants are under‑utilized or overloaded, and decide whether to expand facilities or improve operations.
```

```python
#---------------------------------------------
#Recycling + Composting Rate
#---------------------------------------------

df['Recovery_Rate'] = (df['Recycling_Rate'] + df['Composting_Rate'])
print("Recovery_Rate:",df['Recovery_Rate'])
```

```python
#Combines recycling + composting to show total recovery of waste. Useful for sustainability tracking.
#Helps governments track sustainability progress and decide where to invest — e.g., more composting units or better recycling programs.
```

```python
#---------------------------------------------
#Unprocessed Waste Rate
#---------------------------------------------

df['Collection_Gap_Percentage'] = (100 - df['Collection_Efficiency_Percentage'])
print("Collection_Gap_Percentage:",df['Collection_Gap_Percentage'])
```

```python
#Highlights the percentage of waste not collected. Important for identifying service gaps.
#Municipalities can identify service gaps (areas where waste is left uncollected) and allocate more vehicles or workers to those regions.
```

```python
df["Estimated_Annual_Waste_Tons"] = df["Daily_Waste_Generation_Tons"] * 365
print("Estimated_Annual_Waste_Tons",df["Estimated_Annual_Waste_Tons"])
```

```python
#Projects yearly totals. Helps in long‑term planning and budgeting.
#Useful for long‑term planning — budgeting, landfill capacity, and forecasting future infrastructure needs.
```
